SAIL 2025 (20–24 August) attracted large crowds to the quays and bridges along the IJ. Crowd managers need clear, up-to-date information to see where and when it gets busy. This project uses pedestrian sensor and vessel data from SAIL 2025 to build a monitoring dashboard that supports these decisions.

# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

The objective of this project is to analyse pedestrian flows during SAIL 2025 to identify peak hours, the busiest locations and the vessels that attract the most visitors, and to present these insights in a monitoring dashboard for crowd managers. This supports crowd management during the event and evidence-based planning for SAIL 2030, for example by optimising vessel positioning to spread visitors and applying crowd management measures during peak hours.

1. Which hours of the day have the highest crowd density and when do these numbers reach critical levels (when are peak and off-peak hours)?
2. Which areas of the event are the most visited, and which vessls attract the most visitors?
3. What are the most used routes used by visitors during the event, and how do they change over the time of day, and days?

Expected Impact: Identifies key bottlenecks, highlights areas where residents are most affected, and pinpoints where alternative passages or detours are required.

Pedestrian counts are converted to flow (persons per metre per minute) using the effective width of each sensor location, so that locations of different widths can be compared. Peak hours are defined as the hours in which a sensor's average flow is in the top 25% of its hourly values over the event.

**Possible visualisation**:

1. A map of all vessel location (of the participaing ones)
2. A map of crowd monitor sensors.
3. A map of both the vessel locations and crowd monitors (to observe the amount of ships we can collect information about)
4. Histogram for each crowd monitor (bucket size = 1 hour) to observe the general crowd distribution or line chart
5. Node map with color coded edges based on the busyness level.

**Geographical scale** - The SAIL 2025 event area.

**Temporal scale** - Saturday (10:00-22:00/20:00). If we decide on very frequent time intervals for data aggregation, will not look at other days of the event due to the limitations of streamlit supporting large number of data records at once.

# Data Pipeline

1) Data chunking - filtering data based on the selected date and time
2) Data cleaning – handling missing values and sensor dropouts. Long periods of zero counts during event hours are treated as sensor failures rather than real zeros.
3) Data preparation (+ documentation) – converting counts to flow using the effective width, handling the two directional columns per sensor, and aggregating the data to the selected time interval (e.g. 5 minutes, 1 hour, etc.).
4) Using the data for coding in a unified manner.
